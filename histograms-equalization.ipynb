```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
img=cv2.imread("mountain.jpg")
plt.imshow(img)
plt.axis("off")
```

```python
img=cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.imshow(img,cmap="gray")
plt.axis("off")
```

```python
equ=cv2.equalizeHist(img)
res =np.hstack((img,equ))
plt.imshow(res,cmap="gray")
plt.axis("off")
```
